# Gradient

The gradient of a function $f(x)$ at a point $x$ is a vector that contains all the partial derivatives of the function with respect to each variable. Mathematically, if $x = (x_1, x_2, \ldots, x_n)$, then the gradient $\nabla f(x)$ is given by:

$$\nabla f(\mathbf{x}) = \left( \frac{\partial f}{\partial x_1}, \frac{\partial f}{\partial x_2}, \ldots, \frac{\partial f}{\partial x_n} \right)$$

Each component $\frac{\partial f}{\partial x_i}$ represents the rate of change of the function $f$ with respect to the variable $x_i$. But the gradient as such points towards the maximum of $f$.

### Example

Let's look at a simple example to convince ourselves that the gradient of a function points towards its maximum. The function:

$$f(x, y) = - (x^2 + y^2)$$

has its maximum at the origin (0, 0).

$f$'s gradient can be computed as:

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function
def f(x, y):
    return -(x**2 + y**2)

# Define the gradient of the function
def gradient_f(x, y):
    df_dx = -2 * x
    df_dy = -2 * y
    return np.array([df_dx, df_dy])

# Generate a grid of points
x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 20)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Compute the gradient at each point
U, V = gradient_f(X, Y)

# Plot the function
plt.figure(figsize=(10, 6))
plt.contourf(X, Y, Z, levels=50, cmap='viridis')
plt.colorbar(label='f(x, y)')

# Plot the gradient vectors
plt.quiver(X, Y, U, V, color='white')

# Highlight the maximum point
plt.plot(0, 0, 'ro')  # Maximum point

plt.title('Function and its Gradient')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

## Gradient Points Towards the Maximum

The gradient vector $\nabla f(\mathbf{x})$ points in the direction of the steepest ascent of $f$. In other words, if you move in the direction of the gradient, $f$ will increase most rapidly, and here's why:

1. **Directional Derivative**: The directional derivative of $f$ in the direction of a unit vector $u$ is given by the dot product $\nabla f(x) \cdot u$. This measures the rate of change of $f$ in the direction of $u$

2. **Maximizing the Rate of Increase**: The dot product $\nabla f(x) \cdot u$ is maximized when $u$ is in the same direction as $\nabla f(x)$. This is because the dot product is maximized when the two vectors point in the same direction

Thus, gradient $\nabla f(x)$ points in the direction where the function $f$ increases most rapidly, which is why it points towards the maximum.

## Gradient Descent

In gradient descent, we aim to minimize $f(x)$. To achieve this, we need to move in the direction **opposite** to the gradient. By doing this, we move towards the minimum. Therefore, The update rule in gradient descent is:

$$x_{\text{new}} = x_{\text{old}} - \eta \nabla f(x_{\text{old}})$$

where $\eta$ is the learning rate, a small positive scalar that controls the step size of each update.
